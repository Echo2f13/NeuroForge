# 11 — Additional Information Generation

This notebook demonstrates the Additional Information workflow for NeuroForge.
It generates real-world applications, industry uses, common mistakes, and
interview questions for a given topic using a retrieve → generate → format pipeline.

---

## Overview

The `AdditionalInfoWorkflow` provides supplementary learning material:
1. **Real-world applications** — concrete examples of how the topic is used in practice
2. **Industry uses** — specific sectors and companies that rely on this topic
3. **Common mistakes** — pitfalls learners encounter, with explanations
4. **Interview questions** — questions ranging from basic to advanced difficulty

```python
import sys
sys.path.insert(0, "..")

from src.workflows import AdditionalInfoWorkflow
from src.llm import LLMClient
from src.retrieval import Retriever

print("Imports OK.")
```

## Setup

Initialize the LLM client and retriever. The workflow uses semantic search
for context retrieval and structured JSON output for generation.

```python
# Initialize LLM client (requires API keys in .env)
llm_client = LLMClient()
print(f"Available providers: {llm_client.available_providers}")
```

```python
# Initialize the retriever (requires vector store and knowledge graph)
from src.store import VectorStore, KnowledgeGraph

vector_store = VectorStore()
knowledge_graph = KnowledgeGraph()
retriever = Retriever(vector_store=vector_store, knowledge_graph=knowledge_graph)

print("Retriever initialized.")
```

## Generate Additional Info

Use the workflow to generate supplementary material for a topic.

```python
# Create the workflow
workflow = AdditionalInfoWorkflow(retriever=retriever, llm_client=llm_client)

# Generate additional info for a topic
topic = "neural networks"
result = workflow.generate(topic=topic)

print(f"Generated additional info for: {topic}")
print(f"Keys: {list(result.keys())}")
```

```python
# Display real-world applications
print("\n📱 Real-World Applications:")
print("=" * 40)
for i, app in enumerate(result["applications"], 1):
    print(f"  {i}. {app}")
```

```python
# Display industry uses
print("\n🏭 Industry Uses:")
print("=" * 40)
for i, use in enumerate(result["industry_uses"], 1):
    print(f"  {i}. {use}")
```

```python
# Display common mistakes
print("\n⚠️ Common Mistakes:")
print("=" * 40)
for i, mistake in enumerate(result["common_mistakes"], 1):
    print(f"  {i}. {mistake}")
```

```python
# Display interview questions
print("\n🎯 Interview Questions:")
print("=" * 40)
for i, question in enumerate(result["interview_questions"], 1):
    print(f"  {i}. {question}")
```

## Try Another Topic

The workflow works with any topic in the knowledge base.

```python
# Generate for a different topic
topic_2 = "machine learning"
result_2 = workflow.generate(topic=topic_2)

print(f"\n📚 Additional Info for: {topic_2}")
print("=" * 50)
for key, items in result_2.items():
    print(f"\n  {key} ({len(items)} items):")
    for item in items:
        print(f"    • {item}")
```

## Summary

The `AdditionalInfoWorkflow` provides a simple pipeline for generating
supplementary learning material:

- **Input**: A topic string
- **Pipeline**: Retrieve context → LLM generation → Format output
- **Output**: Dict with applications, industry_uses, common_mistakes, interview_questions

This information helps students understand the practical relevance of
academic topics and prepare for real-world applications and interviews.
